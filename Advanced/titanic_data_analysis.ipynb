```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
sns.set_theme(style="whitegrid")
print("=" * 60)
print("TITANIC DATA ANALYSIS")
print("=" * 60)
df = pd.read_csv("Titanic.csv")
print("\nDataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nFirst 10 rows:")
display(df.head(10))
print("\nDataset Information:")
df.info()
print("\nStatistical Summary:")
display(df.describe())
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())
if df["Age"].isnull().sum() > 0:
    df["Age"] = df["Age"].fillna(df["Age"].median())
if df["Embarked"].isnull().sum() > 0:
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df = df.drop_duplicates()
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
print("\nNumber of duplicate rows after cleaning:",
      df.duplicated().sum())
survival_count = df["Survived"].value_counts()
survival_rate = df["Survived"].mean() * 100
print("\n" + "=" * 60)
print("OVERALL SURVIVAL ANALYSIS")
print("=" * 60)
print("\nSurvival Count:")
print(survival_count)
print("\nOverall Survival Rate:",
      round(survival_rate, 2), "%")
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Survived")
plt.title("Titanic Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
gender_survival = df.groupby("Sex")["Survived"].mean() * 100
print("\n" + "=" * 60)
print("GENDER ANALYSIS")
print("=" * 60)
print("\nSurvival Rate by Gender:")
print(gender_survival.round(2))
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
plt.figure(figsize=(7, 5))
gender_survival.plot(kind="bar")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
class_survival = df.groupby("Pclass")["Survived"].mean() * 100
print("\n" + "=" * 60)
print("PASSENGER CLASS ANALYSIS")
print("=" * 60)
print("\nSurvival Rate by Passenger Class:")
print(class_survival.round(2))
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Pclass", hue="Survived")
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
plt.figure(figsize=(7, 5))
class_survival.plot(kind="bar")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
print("\n" + "=" * 60)
print("AGE ANALYSIS")
print("=" * 60)
print("\nAverage Age:",
      round(df["Age"].mean(), 2))
print("Minimum Age:",
      df["Age"].min())
print("Maximum Age:",
      df["Age"].max())
plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    kde=True
)
plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=[
        "Child",
        "Teenager",
        "Young Adult",
        "Adult",
        "Senior"
    ]
)
age_group_survival = (
    df.groupby("AgeGroup", observed=False)["Survived"].mean() * 100
)
print("\nSurvival Rate by Age Group:")
print(age_group_survival.round(2))
plt.figure(figsize=(8, 5))
sns.barplot(
    x=age_group_survival.index,
    y=age_group_survival.values
)
plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=20)
plt.show()
print("\n" + "=" * 60)
print("FARE ANALYSIS")
print("=" * 60)
print("\nAverage Fare:",
      round(df["Fare"].mean(), 2))
print("Minimum Fare:",
      df["Fare"].min())
print("Maximum Fare:",
      df["Fare"].max())
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="Survived",
    y="Fare"
)
plt.title("Fare Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Fare")
plt.show()
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
family_survival = (
    df.groupby("FamilySize")["Survived"].mean() * 100
)
print("\n" + "=" * 60)
print("FAMILY SIZE ANALYSIS")
print("=" * 60)
print("\nSurvival Rate by Family Size:")
print(family_survival.round(2))
plt.figure(figsize=(9, 5))
sns.barplot(
    x=family_survival.index,
    y=family_survival.values
)
plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Survival Rate (%)")
plt.show()
numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()
print("\n" + "=" * 60)
print("CORRELATION MATRIX")
print("=" * 60)
display(correlation.round(2))
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
print("\n" + "=" * 60)
print("FINAL PROJECT SUMMARY")
print("=" * 60)
print("\nTotal Passengers:",
      len(df))
print(
    "Overall Survival Rate:",
    round(df["Survived"].mean() * 100, 2),
    "%"
)
print("\nSurvival Rate by Gender:")
print(
    (df.groupby("Sex")["Survived"].mean() * 100).round(2)
)
print("\nSurvival Rate by Passenger Class:")
print(
    (df.groupby("Pclass")["Survived"].mean() * 100).round(2)
)
print("\nAverage Age:",
      round(df["Age"].mean(), 2))
print("Average Fare:",
      round(df["Fare"].mean(), 2))
print("\n" + "=" * 60)
print("KEY FINDINGS")
print("=" * 60)

print("""
1. The overall survival rate was calculated from the dataset.

2. Survival rates differed between male and female passengers.

3. Passenger class showed a relationship with survival.

4. Age distributions differed between survivors and
   non-survivors.

5. Fare values showed differences between survival groups.

6. Family size was also analyzed to understand its
   relationship with survival.

7. The correlation heatmap helped identify relationships
   between numerical variables.

8. Data visualization made the patterns in the dataset
   easier to understand.
""")
df.to_csv("Titanic_Cleaned.csv", index=False)
print("\nCleaned dataset saved as: Titanic_Cleaned.csv")
print("\n" + "=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
